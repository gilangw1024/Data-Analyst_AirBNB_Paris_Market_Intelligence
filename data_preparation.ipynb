```python
import pandas as pd
import numpy as np
```

```python
# Contoh membaca file CSV
df = pd.read_csv('dataset/airbnb_listing.csv')

# Cek 5 baris pertama data
df.head()
```

```python
df.info()
```

```python
#isi bedroom, bed, dan guests yang memiliki missing value dengan median
df['guests'] = df.groupby(['room_type', 'property_type'])['guests'].transform(lambda x: x.fillna(x.median()))
df['bedrooms'] = df.groupby(['room_type', 'property_type'])['bedrooms'].transform(lambda x: x.fillna(x.median()))
df['beds'] = df.groupby(['room_type', 'property_type'])['beds'].transform(lambda x: x.fillna(x.median())) 

#Apa yang dilakukan kode tersebut?
# Data dikelompokkan berdasarkan kombinasi room_type + property_type
# Di dalam setiap kelompok, dihitung median dari kolom guests
# Missing value di dalam kelompok tersebut diisi dengan median kelompoknya
# Hasilnya dikembalikan ke kolom guests menggunakan .transform()
```

```python
# pada kondisi lain, langsung diisi langsung dengan median
df['guests'] = df['guests'].fillna(df['guests'].median())
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())
df['beds'] = df['beds'].fillna(df['beds'].median())
```

```python
#handling superhost, neighbor, dan star rating
df['superhost'] = df['superhost'].fillna(False).astype(bool)
df['neighborhood'] = df['neighborhood'].fillna('Unknown')
df['star_rating'] = df['star_rating'].fillna(0.0)
```

```python
#data type dan parsing
df['month_date'] = pd.to_datetime(df['month_date'])
df['listing_id'] = df['listing_id'].astype(str)
df['host_id'] = df['host_id'].astype(str)
```

```python
#ubah ke interger setelah data diinputasi
df['guests'] = df['guests'].astype(int)
df['bedrooms'] = df['bedrooms'].astype(int)
df['beds'] = df['beds'].astype(int)
```
